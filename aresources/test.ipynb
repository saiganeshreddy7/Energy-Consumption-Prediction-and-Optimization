```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# Set random seed for reproducibility
np.random.seed(42)

# Number of rows
num_rows = 10000

# Generate synthetic data
timestamps = pd.date_range(start='2025-01-01', periods=num_rows, freq='H')
energy_consumption_kwh = np.random.normal(100, 20, num_rows)
peak_demand_kw = np.random.normal(50, 10, num_rows)
outdoor_temperature_c = np.random.normal(25, 5, num_rows)
humidity_percent = np.random.normal(50, 10, num_rows)
solar_radiation_w_per_m2 = np.random.normal(500, 200, num_rows)
wind_speed_m_per_s = np.random.normal(5, 2, num_rows)
occupancy_level = np.random.randint(0, 100, num_rows)
hvac_settings_c = np.random.normal(22, 2, num_rows)
lighting_usage_kwh = np.random.randint(0, 2, num_rows)  # Assuming binary on/off state
appliance_power_consumption_kwh = np.random.normal(2, 1, num_rows)
battery_storage_level_percent = np.random.randint(0, 100, num_rows)
grid_energy_price_per_kwh = np.random.normal(0.15, 0.05, num_rows)
renewable_energy_usage_percent = np.random.uniform(0, 50, num_rows)

# Create DataFrame
data = {
    'Timestamp': timestamps,
    'Energy Consumption (kWh)': energy_consumption_kwh,
    'Peak Demand (kW)': peak_demand_kw,
    'Outdoor Temperature (°C)': outdoor_temperature_c,
    'Humidity (%)': humidity_percent,
    'Solar Radiation (W/m²)': solar_radiation_w_per_m2,
    'Wind Speed (m/s)': wind_speed_m_per_s,
    'Occupancy Level': occupancy_level,
    'HVAC Settings (°C)': hvac_settings_c,
    'Lighting Usage (On/Off or kWh)': lighting_usage_kwh,
    'Appliance Power Consumption (kWh)': appliance_power_consumption_kwh,
    'Battery Storage Level (%)': battery_storage_level_percent,
    'Grid Energy Price ($/kWh)': grid_energy_price_per_kwh,
    'Renewable Energy Usage (%)': renewable_energy_usage_percent
}

df = pd.DataFrame(data)

# Display first few rows of the generated data
print(df.head())

# Save data to a CSV file if needed
df.to_csv('energy_consumption_data_10000.csv', index=False)
```
